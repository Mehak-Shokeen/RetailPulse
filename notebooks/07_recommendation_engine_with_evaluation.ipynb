# Module 4 — RetailPulse Recommendation Engine

## Objective
Build a basket-based product recommender using the association rules exported by Module 3. Given one or more products already in a basket, the engine finds rules whose antecedents are fully present and ranks new product suggestions using **confidence, lift, and support**.

This is an association-rule recommender, not a customer-personalisation model. It does not use customer identity, segmentation labels, or inactivity scores in this module.

### Inputs
- `artifacts/association/association_rules.csv` — generated by Module 3.
- `data/processed/cleaned_transactions.csv` — used only to map stock codes to readable product descriptions.

### Outputs
- `artifacts/recommendation/recommendation_engine_sample.csv` — example recommendations from a reproducible sample basket.
- `artifacts/recommendation/recommendation_engine_metadata.json` — input/output details and recommendation logic.

## 1. Imports and project paths
Run this notebook from the RetailPulse project. The path helper checks the current directory and its parents so it also works when Jupyter starts inside `notebooks/`.

```python
from pathlib import Path
import json
import pandas as pd
import numpy as np
from IPython.display import display


def find_project_root(start: Path) -> Path:
    for candidate in [start, *start.parents]:
        if (candidate / "data" / "processed" / "cleaned_transactions.csv").exists() and (candidate / "artifacts" / "association").exists():
            return candidate
    raise FileNotFoundError("Could not locate RetailPulse project root. Run this notebook inside the RetailPulse project.")

PROJECT_ROOT = find_project_root(Path.cwd())
RULES_PATH = PROJECT_ROOT / "artifacts" / "association" / "association_rules.csv"
TRANSACTIONS_PATH = PROJECT_ROOT / "data" / "processed" / "cleaned_transactions.csv"
OUTPUT_DIR = PROJECT_ROOT / "artifacts" / "recommendation"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Project root:", PROJECT_ROOT)
print("Association rules exist:", RULES_PATH.exists())
print("Transactions file exists:", TRANSACTIONS_PATH.exists())
print("Output directory:", OUTPUT_DIR)
```

## 2. Load Module 3 rules and product descriptions
Stock codes are read as strings to preserve identifiers. Product descriptions are used for display only; matching is performed on stock codes.

```python
if not RULES_PATH.exists():
    raise FileNotFoundError(f"Missing {RULES_PATH}. Run Module 3 and export association_rules.csv first.")
if not TRANSACTIONS_PATH.exists():
    raise FileNotFoundError(f"Missing {TRANSACTIONS_PATH}.")

rules = pd.read_csv(RULES_PATH, dtype={"antecedent_codes": "string", "consequent_codes": "string"})
product_data = pd.read_csv(TRANSACTIONS_PATH, usecols=["StockCode", "Description"], dtype={"StockCode": "string"})

required_rule_columns = {"antecedent_codes", "consequent_codes", "support", "confidence", "lift"}
missing_columns = required_rule_columns - set(rules.columns)
if missing_columns:
    raise ValueError(f"Association rules file is missing columns: {sorted(missing_columns)}")

product_data = product_data.dropna(subset=["StockCode", "Description"]).copy()
product_data["StockCode"] = product_data["StockCode"].str.strip()
product_data["Description"] = product_data["Description"].str.strip()
product_data = product_data[product_data["StockCode"].ne("")]
product_data = product_data.drop_duplicates(subset="StockCode", keep="first")
code_to_name = dict(zip(product_data["StockCode"], product_data["Description"]))
name_to_codes = {}
for stock_code, description in code_to_name.items():
    name_to_codes.setdefault(description.casefold(), []).append(stock_code)

print(f"Association rules loaded: {len(rules):,}")
print(f"Product descriptions mapped: {len(code_to_name):,}")
display(rules.head())
```

## 3. Prepare rule antecedents and consequents
The exported CSV stores itemsets as pipe-separated stock codes. Convert these strings into sets so basket matching is reliable.

```python
def parse_item_codes(value):
    if pd.isna(value):
        return frozenset()
    return frozenset(part.strip() for part in str(value).split("|") if part.strip())

rules = rules.copy()
rules["antecedent_set"] = rules["antecedent_codes"].apply(parse_item_codes)
rules["consequent_set"] = rules["consequent_codes"].apply(parse_item_codes)
for metric in ["support", "confidence", "lift"]:
    rules[metric] = pd.to_numeric(rules[metric], errors="coerce")
rules = rules.dropna(subset=["support", "confidence", "lift"])
rules = rules[(rules["antecedent_set"].map(len) > 0) & (rules["consequent_set"].map(len) > 0)]
print(f"Usable rules: {len(rules):,}")
print("Antecedent sizes:")
display(rules["antecedent_set"].map(len).value_counts().sort_index().rename_axis("number_of_products").to_frame("rules"))
```

## 4. Product input resolver
The recommender accepts either stock codes or exact product descriptions. If a description maps to multiple stock codes, it asks for the stock code to avoid silently choosing the wrong product.

```python
def resolve_product(value):
    value = str(value).strip()
    if value in code_to_name:
        return value
    matches = name_to_codes.get(value.casefold(), [])
    if len(matches) == 1:
        return matches[0]
    if len(matches) > 1:
        raise ValueError(f"Description is ambiguous; use a StockCode instead: {value} -> {matches[:10]}")
    raise ValueError(f"Product not found by StockCode or exact Description: {value}")


def product_label(stock_code):
    return code_to_name.get(stock_code, "Unknown product")
```

## 5. Recommendation function

For an input basket, a rule is eligible when **all products in its antecedent are already in the basket**. Products already in the basket are removed from the suggestions. If multiple rules suggest the same product, the best available confidence/lift/support values are retained for ranking.

Ranking is lexicographic: confidence first, then lift, then support. These metrics are displayed separately so the recommendation remains interpretable.

```python
def recommend_products(basket, top_n=5):
    """Recommend products from association rules for a basket of stock codes or exact descriptions."""
    if top_n < 1:
        raise ValueError("top_n must be at least 1")
    if isinstance(basket, (str, int, float)):
        basket = [basket]
    basket_codes = {resolve_product(item) for item in basket}
    if not basket_codes:
        raise ValueError("Provide at least one product in the basket.")

    matched = rules[rules["antecedent_set"].apply(lambda items: items.issubset(basket_codes))].copy()
    if matched.empty:
        return pd.DataFrame(columns=["recommended_stock_code", "recommended_product", "confidence", "lift", "support", "matched_antecedent"])

    rows = []
    for _, rule in matched.iterrows():
        for candidate in rule["consequent_set"] - basket_codes:
            rows.append({
                "recommended_stock_code": candidate,
                "confidence": float(rule["confidence"]),
                "lift": float(rule["lift"]),
                "support": float(rule["support"]),
                "matched_antecedent": " | ".join(sorted(rule["antecedent_set"]))
            })

    if not rows:
        return pd.DataFrame(columns=["recommended_stock_code", "recommended_product", "confidence", "lift", "support", "matched_antecedent"])

    candidates = pd.DataFrame(rows)
    candidates = candidates.sort_values(
        ["confidence", "lift", "support"], ascending=False
    ).drop_duplicates(subset="recommended_stock_code", keep="first")
    candidates["recommended_product"] = candidates["recommended_stock_code"].map(product_label)
    columns = ["recommended_stock_code", "recommended_product", "confidence", "lift", "support", "matched_antecedent"]
    return candidates[columns].head(top_n).reset_index(drop=True)
```

## 6. Run a reproducible example
Use the antecedent of the highest-lift rule as a demonstration basket. This guarantees that the example basket is represented in the mined rule set. It is a technical demonstration, not an evaluation on unseen customer data.

```python
best_rule = rules.sort_values(["lift", "confidence", "support"], ascending=False).iloc[0]
sample_basket_codes = sorted(best_rule["antecedent_set"])
print("Example basket:")
for code in sample_basket_codes:
    print(f"- {code}: {product_label(code)}")

sample_recommendations = recommend_products(sample_basket_codes, top_n=10)
print("\nRecommended products:")
display(sample_recommendations)

assert not set(sample_recommendations["recommended_stock_code"]).intersection(sample_basket_codes)
print("Validation passed: no recommended item is already in the input basket.")
```

## 9. Interpretation and limitations

- A recommendation is produced because an association rule's antecedent is contained in the current basket; this indicates co-occurrence, not causation.
- Confidence estimates how often the consequent appeared when the antecedent appeared in the mined transactions.
- Lift compares that confidence with the consequent's overall frequency. Lift greater than 1 indicates a positive association.
- Support indicates the proportion of invoice baskets containing the complete rule itemset.
- The current recommender is basket-based. It does not yet personalise suggestions by customer profile, segment, inactivity probability, or time.
- If no rule matches the supplied basket, the function returns an empty result rather than inventing recommendations.
- The example generated here demonstrates function behaviour; it is not a formal offline accuracy evaluation. A separate train/test evaluation would be needed before claiming recommendation quality.

## 10. Offline evaluation on later invoices (temporal holdout)

The earlier example demonstrates that the recommender runs; it is not an accuracy test. This section performs a separate offline **basket-completion evaluation**. It trains association rules using only earlier invoices, then tests whether those rules can recover products deliberately held out from later invoices.

For each eligible test invoice, products are split deterministically into an input context basket and a held-out target basket. This simulates a customer who has already placed some products in a basket and may still add others. It is an offline proxy, not a live A/B test or a customer-personalised evaluation.

Metrics:
- **Precision@K:** relevant held-out products among the top K recommendations, divided by the number of recommendations returned (for invoices with recommendations).
- **Recall@K:** relevant held-out products among the top K recommendations, divided by the number of held-out products.
- **Coverage@K:** eligible test invoices receiving at least one recommendation in the top K, divided by all eligible test invoices.

To avoid temporal leakage, the rules used here are re-mined from the training period only; the full-data Module 3 rules are not used for this evaluation.


```python
from mlxtend.frequent_patterns import fpgrowth, association_rules
from scipy.sparse import csr_matrix
from collections import defaultdict

# Evaluation settings. Keep aligned with Module 3 where possible.
TRAIN_FRACTION = 0.80
EVAL_MIN_SUPPORT = 0.01
EVAL_MIN_CONFIDENCE = 0.20
EVAL_TOP_K = [5, 10]
RANDOM_SEED = 42

# Load only the columns required for invoice-level basket evaluation.
eval_data = pd.read_csv(
    TRANSACTIONS_PATH,
    usecols=["Invoice", "StockCode", "InvoiceDate", "Quantity", "Price"],
    dtype={"Invoice": "string", "StockCode": "string"},
    parse_dates=["InvoiceDate"]
)

# Keep valid positive purchase lines. This mirrors the cleaned purchase-line
# definition used in Module 3 and excludes cancellation/return lines if present.
eval_data["StockCode"] = eval_data["StockCode"].str.strip()
eval_data = eval_data.dropna(subset=["Invoice", "StockCode", "InvoiceDate"])
eval_data = eval_data[
    (eval_data["StockCode"] != "") &
    (eval_data["Quantity"] > 0) &
    (eval_data["Price"] > 0)
].copy()

# Aggregate to one timestamp per invoice, then split chronologically.
invoice_dates = eval_data.groupby("Invoice", observed=True)["InvoiceDate"].min().sort_values()
cut_position = int(len(invoice_dates) * TRAIN_FRACTION)
cutoff_date = invoice_dates.iloc[cut_position]
train_invoice_ids = set(invoice_dates[invoice_dates < cutoff_date].index)
test_invoice_ids = set(invoice_dates[invoice_dates >= cutoff_date].index)

print(f"Temporal cutoff: {cutoff_date}")
print(f"Training invoices: {len(train_invoice_ids):,}")
print(f"Test invoices: {len(test_invoice_ids):,}")
```

```python
# Build training baskets. A product appears at most once per invoice.
train_lines = eval_data[eval_data["Invoice"].isin(train_invoice_ids)]
train_baskets = (
    train_lines.groupby("Invoice", observed=True)["StockCode"]
    .apply(lambda values: sorted(set(values)))
    .tolist()
)

# Build a sparse invoice-product matrix to avoid allocating a dense 0/1 matrix.
train_products = sorted({product for basket in train_baskets for product in basket})
product_to_col = {product: i for i, product in enumerate(train_products)}
row_indices, col_indices = [], []
for row_idx, basket in enumerate(train_baskets):
    for product in basket:
        row_indices.append(row_idx)
        col_indices.append(product_to_col[product])

train_matrix = csr_matrix(
    (np.ones(len(row_indices), dtype=bool), (row_indices, col_indices)),
    shape=(len(train_baskets), len(train_products)),
    dtype=bool
)
train_encoded = pd.DataFrame.sparse.from_spmatrix(
    train_matrix,
    index=range(len(train_baskets)),
    columns=train_products
)

print(f"Training basket matrix: {train_encoded.shape[0]:,} invoices × {train_encoded.shape[1]:,} products")
print(f"Training basket density: {train_matrix.nnz / (train_matrix.shape[0] * train_matrix.shape[1]):.6f}")
```

```python
# Re-mine rules using TRAINING DATA ONLY. This prevents the later test invoices
# from influencing the rules being evaluated.
eval_itemsets = fpgrowth(
    train_encoded,
    min_support=EVAL_MIN_SUPPORT,
    use_colnames=True,
    max_len=3
)

if eval_itemsets.empty:
    raise ValueError("No frequent itemsets found for the evaluation training period. Consider reviewing the support threshold.")

eval_rules = association_rules(
    eval_itemsets,
    metric="confidence",
    min_threshold=EVAL_MIN_CONFIDENCE
)

eval_rules = eval_rules[eval_rules["lift"] > 1].copy()
eval_rules["antecedent_set"] = eval_rules["antecedents"].apply(frozenset)
eval_rules["consequent_set"] = eval_rules["consequents"].apply(frozenset)
eval_rules = eval_rules.sort_values(
    ["confidence", "lift", "support"], ascending=False
).reset_index(drop=True)

print(f"Training frequent itemsets: {len(eval_itemsets):,}")
print(f"Training association rules (confidence ≥ {EVAL_MIN_CONFIDENCE}, lift > 1): {len(eval_rules):,}")
```

```python
def recommend_from_eval_rules(input_codes, top_n=10):
    """Return top-N products using rules mined exclusively on training invoices."""
    input_codes = set(input_codes)
    candidates = []
    for _, rule in eval_rules.iterrows():
        if rule["antecedent_set"].issubset(input_codes):
            for product in rule["consequent_set"] - input_codes:
                candidates.append((
                    product,
                    float(rule["confidence"]),
                    float(rule["lift"]),
                    float(rule["support"])
                ))
    if not candidates:
        return []
    candidate_frame = pd.DataFrame(
        candidates, columns=["product", "confidence", "lift", "support"]
    )
    candidate_frame = candidate_frame.sort_values(
        ["confidence", "lift", "support"], ascending=False
    ).drop_duplicates("product", keep="first")
    return candidate_frame["product"].head(top_n).tolist()

# Create deterministic partial-basket contexts from later invoices.
# Only products known during training are eligible for recommendation/evaluation.
rng = np.random.default_rng(RANDOM_SEED)
test_baskets = (
    eval_data[eval_data["Invoice"].isin(test_invoice_ids)]
    .groupby("Invoice", observed=True)["StockCode"]
    .apply(lambda values: sorted(set(values)))
)

evaluation_cases = []
for invoice_id, full_basket in test_baskets.items():
    known_products = [p for p in full_basket if p in product_to_col]
    if len(known_products) < 2:
        continue
    shuffled = list(known_products)
    rng.shuffle(shuffled)
    split_at = max(1, len(shuffled) // 2)
    context = set(shuffled[:split_at])
    heldout = set(shuffled[split_at:])
    if context and heldout:
        evaluation_cases.append((invoice_id, context, heldout))

print(f"Eligible held-out test invoices: {len(evaluation_cases):,}")
```

```python
# Compute Precision@K, Recall@K and Coverage@K.
metric_rows = []
for k in EVAL_TOP_K:
    precision_values = []
    recall_values = []
    covered = 0

    for _, context, heldout in evaluation_cases:
        recommendations = recommend_from_eval_rules(context, top_n=k)
        hits = len(set(recommendations) & heldout)
        precision_values.append(hits / k)
        recall_values.append(hits / len(heldout))
        if recommendations:
            covered += 1

    total_cases = len(evaluation_cases)
    metric_rows.append({
        "K": k,
        "Precision@K": float(np.mean(precision_values)) if precision_values else 0.0,
        "Recall@K": float(np.mean(recall_values)) if recall_values else 0.0,
        "Coverage@K": covered / total_cases if total_cases else 0.0,
        "Eligible_test_invoices": total_cases,
        "Invoices_with_recommendations": covered
    })

evaluation_results = pd.DataFrame(metric_rows)
display(evaluation_results.style.format({
    "Precision@K": "{:.4f}",
    "Recall@K": "{:.4f}",
    "Coverage@K": "{:.2%}"
}))
```

```python
# ============================================================
# EXPORT RECOMMENDATION EVALUATION RESULTS
# ============================================================

from pathlib import Path
import json
import pandas as pd

# 1. Create output directory
evaluation_output_dir = (
    Path(PROJECT_ROOT) / "artifacts" / "recommendation"
)
evaluation_output_dir.mkdir(parents=True, exist_ok=True)

# 2. Define output paths
evaluation_csv_path = (
    evaluation_output_dir / "recommendation_evaluation.csv"
)

evaluation_rules_path = (
    evaluation_output_dir / "evaluation_association_rules.csv"
)

evaluation_metadata_path = (
    evaluation_output_dir / "evaluation_metadata.json"
)

# 3. Validate evaluation results
required_columns = [
    "K",
    "Precision@K",
    "Recall@K",
    "Coverage@K",
    "Eligible_test_invoices",
    "Invoices_with_recommendations"
]

missing_columns = [
    col for col in required_columns
    if col not in evaluation_results.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns in evaluation_results: {missing_columns}\n"
        f"Available columns: {list(evaluation_results.columns)}"
    )

# 4. Export evaluation metrics
evaluation_results.to_csv(
    evaluation_csv_path,
    index=False
)

# 5. Prepare training-only association rules for export
export_eval_rules = eval_rules.drop(
    columns=[
        "antecedents",
        "consequents",
        "antecedent_set",
        "consequent_set"
    ],
    errors="ignore"
).copy()

export_eval_rules["antecedents"] = (
    eval_rules["antecedent_set"].apply(
        lambda x: " | ".join(sorted(map(str, x)))
    )
)

export_eval_rules["consequents"] = (
    eval_rules["consequent_set"].apply(
        lambda x: " | ".join(sorted(map(str, x)))
    )
)

export_eval_rules.to_csv(
    evaluation_rules_path,
    index=False
)

# 6. Extract invoice counts directly from evaluation results
eligible_test_invoices = int(
    evaluation_results["Eligible_test_invoices"].iloc[0]
)

invoices_with_recommendations = int(
    evaluation_results["Invoices_with_recommendations"].iloc[0]
)

# 7. Prepare evaluation metadata
evaluation_metadata = {
    "evaluation_type": "temporal_holdout",
    "metrics": json.loads(
        evaluation_results.to_json(orient="records")
    ),
    "eligible_test_invoices": eligible_test_invoices,
    "invoices_with_recommendations": invoices_with_recommendations,
    "evaluation_scope": (
        "Recommendations evaluated against later purchases"
    )
}

# 8. Save metadata
evaluation_metadata_path.write_text(
    json.dumps(evaluation_metadata, indent=2),
    encoding="utf-8"
)

# 9. Confirm successful exports
print("Evaluation exports completed successfully!\n")

print("Saved files:")

for path in [
    evaluation_csv_path,
    evaluation_rules_path,
    evaluation_metadata_path
]:
    print("-", path.relative_to(PROJECT_ROOT))

print("\nEvaluation summary:")
print(evaluation_results.to_string(index=False))
```

## 11. How to interpret the evaluation

- Precision@K measures the proportion of the returned top-K suggestions that appear in the held-out portion of the invoice.
- Recall@K measures the proportion of held-out products recovered by the top-K suggestions.
- Coverage@K measures how often the system can produce at least one recommendation for an eligible test invoice.
- The evaluation uses rules mined from earlier invoices and later invoices for testing. The input/target split within each test invoice is a controlled basket-completion simulation.
- Results depend on the chronological cutoff, support/confidence thresholds, and the fact that only products observed in the training period can be recommended.
- These metrics should be reported as offline evaluation results, not as evidence of uplift in real customer purchases.
